```python
import cv2
```

```python
body_classifier = cv2.CascadeClassifier(r"C:\Users\LoQ\Desktop\AI\(AI) Deep learning\CV\haarcascade_fullbody.xml")
```

```python
cap = cv2.VideoCapture(r"C:\Users\LoQ\Desktop\AI\(AI) Deep learning\CV\walking.avi")
```

```python
while cap.isOpened():
    check ,frame =cap.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    bodies= body_classifier.detectMultiScale(gray,1.2,3)

    for (x,y,w,h) in bodies:
        cv2.rectangle(frame,(x,y),(x+w,y+h), (0,255,255),2)
        cv2.imshow('Pedestrains',frame)
    if cv2.waitKey(1)==ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
